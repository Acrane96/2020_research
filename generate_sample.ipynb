```python
import csv

infilename = r'./data/nf-chunk2.csv'
outfilename = r'./data/sample.csv'

with open(infilename, 'r') as fp_in, open(outfilename, 'a') as fp_out:
    reader = csv.reader(fp_in, delimiter=",")
    # headers = next(reader)  # read first row
    row_count = 20000000
    writer = csv.writer(fp_out, delimiter=",")
    i = 0
    for row in reader:
        i+=1
        writer.writerow(row)
        if i > row_count/4:
            print("i: ", i, "row_count/4: ", row_count/4)
            break
```
